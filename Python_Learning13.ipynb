# 👥  Object-Oriented Programming: Classes and Objects

Python is an **Object-Oriented Programming (OOP)** language. Almost everything in Python is an object, with its properties and methods.

📌 **Core Concept Definitions:**
* **Class**: A blueprint, template, or prototype used to define the structure and behavior of an object. It defines what attributes and actions the object will have.
* **Object**: A concrete instance created from a class blueprint. While a class is a conceptual plan, an object is the actual entity that occupies system memory.

📁 **Syntax Structure:**
```python
class ClassName:
    # Class attributes and methods go here
    pass
```


````python
class NewClass:
    num = 10

# print(NewClass)
p1 = NewClass()
print(p1.num)
````

````python
# Define a simple blueprint (Class) for a car
class CarBlueprint:
    # A simple class attribute shared by all instances
    vehicle_type = "Automobile"

# Instantiate an actual object from the class template
my_car_object = CarBlueprint()

# Access the class property using dot notation
print(f"Object Type: {type(my_car_object)}")
print(f"Shared Attribute: {my_car_object.vehicle_type}")
````

### 👥 1. Object Initializers and the `__init__()` Method

To assign data values to an object at the moment it is created, use the built-in constructor method named `__init__()`. 

📌 **Important Technical Details:**
* The `__init__()` function runs automatically every time a new object is initiated from the class.
* The `self` parameter is a mandatory reference to the current instance of the class. It allows you to bind variables to the specific object being built.


````python
# Using init
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

ob1 = Person("Dony", 40)

print(ob1.name)
print(ob1.age)
````

````python
# Using init
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def hi(self):
        print("Hi, my name is " + self.name)

ob1 = Person("Dony", 40)
ob1.hi()
````

````python
# Define a class with an initialization constructor
class Vehicle:
    def __init__(self, brand, model, year):
        # Bind incoming arguments to the specific object instance
        self.brand = brand
        self.model = model
        self.year = year

# Create two entirely separate, independent objects from the same class template
car1 = Vehicle("BMW", "X5", 2021)
car2 = Vehicle("Tesla", "Model Y", 2024)

# Print specific object instance attributes
print(f"Car 1 Details: {car1.year} {car1.brand} {car1.model}")
print(f"Car 2 Details: {car2.year} {car2.brand} {car2.model}")
````

### ⚙️ 2. Modifying Object Properties

Once an object is created, its internal attributes can be directly updated or modified at any point in your script using dot notation assignment (`object.attribute = new_value`).


````python
# Define a class pattern based on the notebook workspace sample
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
        
    def hi(self):
        print("Hi, my name is " + self.name)

# Create an object instance
obj1 = Person("Dony", 40)

# Modify the age attribute dynamically
obj1.age = 50

# Print the modified property value to verify the change
print(obj1.age)
````

### 🗑️ 3. Deleting Object Properties and Variables

You can use the built-in `del` keyword to completely wipe out properties from an object or to destroy the entire object container itself from memory. Trying to read a property after running `del` on it will throw a traceback crash.

⚠️ **Expected Error Behavior:**
Attempting to reference a deleted attribute results in an `AttributeError`, letting you know that the object has no property with that name.


````python
# Re-instantiate the Person class structure from the example image
obj1 = Person("Dony", 40)

# 1. Modify the age attribute
obj1.age = 50

# 2. Use the del keyword to delete the specific object property
del obj1.age

# 3. Trying to access obj1.age now will raise an AttributeError trace
try:
    print(obj1.age)
except AttributeError as error_msg:
    print(f"AttributeError Traceback: {error_msg}")
````

### 🗑️ 4. Deleting an Entire Object Instance

Similarly, you can apply the `del` keyword directly to the object variable container itself. This cleans up system space by removing the entire instantiation reference from local execution scope.


````python
# Completely destroy the entire object variable container
del obj1

# Trying to call the variable now raises a NameError because it no longer exists
try:
    print(obj1)
except NameError as error_msg:
    print(f"NameError Traceback: {error_msg}")
````
