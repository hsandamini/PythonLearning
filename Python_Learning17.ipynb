# 🗄️ Introduction to SQLite

**SQLite** is a lightweight, disk-based relational database management system (RDBMS) that stores data in a single file on your system.

📌 **Core Ecosystem Features:**
* **Python `sqlite3` Integration**: Python includes a built-in library module named `sqlite3` right out of the box, meaning you do not need to run any external install commands to build database connections.
* **Relational Database Structure**: It uses relational database principles, structuring data into formal tables containing rows and columns that can be filtered and retrieved using standard Structured Query Language (SQL) statements.


### 🔗 1. Connecting to an SQLite Database and Creating a Cursor

To interact with an SQLite database, you must first establish a connection link using the `sqlite3.connect()` function. If the specified database file does not exist, Python automatically creates it. After connecting, you generate a **cursor** object to execute your Structured Query Language (SQL) statements.


```python
import sqlite3

# Establish a connection to a database file
conn = sqlite3.connect('my_database.db')

# Create a cursor object to execute SQL commands
cursor = conn.cursor()

print("Database connection established successfully.")
```

### 🏗️ 2. Creating a Table and Specifying Data Fields

Data inside relational databases is stored within structured tables. You create tables by executing a `CREATE TABLE` statement via your cursor, specifying column names along with their strict data types (such as `TEXT`, `INTEGER`, or `REAL`).

```python
# Execute an SQL command to create a new table if it does not already exist
cursor.execute('''
    CREATE TABLE IF NOT EXISTS users (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        name TEXT NOT NULL,
        age INTEGER
    )
''')

# Commit the changes to save the structure permanently
conn.commit()

print("Table structure created successfully.")
```

### 🔒 3. Inserting Data and Closing the Connection

To add records to your table, execute an `INSERT INTO` statement. Always use `conn.commit()` after data updates to save your changes to the disk, and remember to use `.close()` on both your cursor and connection objects to prevent data corruption and release system memory.

```python
import sqlite3

# 1. Establish database connection file handler link
conn = sqlite3.connect('my_database.db')
cursor = conn.cursor()

# 2. Automatically verify and build the database table structure if missing
cursor.execute('''
    CREATE TABLE IF NOT EXISTS users (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        name TEXT NOT NULL,
        age INTEGER
    )
''')
conn.commit()

# 3. Securely append your target data records into the verified table
cursor.execute("INSERT INTO users (name, age) VALUES ('Alice', 30)")
cursor.execute("INSERT INTO users (name, age) VALUES ('Bob', 25)")

# 4. Save structural transactions directly to disk memory
conn.commit()

# 5. Cleanly terminate connection resources to release file system locks
cursor.close()
conn.close()

print("Database table verified, records written, and connections closed successfully.")
```

### 🔌 4. Connect to Database with Python

To interact with an SQLite database file from Python, you use the built-in `sqlite3.connect()` function. This opens a connection channel to the file. If the file name you pass does not exist on your computer, Python creates a new, blank database file with that name automatically.

```python
import sqlite3

# Connect to a database file (creates it if missing)
conn = sqlite3.connect('learning_data.db')

print("Database connection opened successfully.")
```

### 🏗️ 5. Create Table in Database using Python

To execute SQL commands, you must first create a **cursor** object from your connection. Tables are built by running a `CREATE TABLE` statement through the cursor, where you define the names of your columns and their corresponding database data types.

```python
# Create a cursor object
cursor = conn.cursor()

# Run SQL statement to create a table structure
cursor.execute('''
    CREATE TABLE IF NOT EXISTS students (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        name TEXT NOT NULL,
        grade REAL
    )
''')

# Commit the changes to save the table to disk
conn.commit()
print("Table structure initialized.")
```
