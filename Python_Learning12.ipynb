# 🔨 Functions in Python

### 🔨 1. Defining Functions
To create a function, write the **`def`** keyword followed by a unique function name and parentheses `()`.
* **The Colon (`:`):** Declares the start of the code block.
* **Indentation:** The action code underneath **must** be indented by 4 spaces.
* **Execution:** Defining a function does not run it. You must explicitly **call** the function using its name with parentheses.

```python
# 1. Defining the function structure
def greet_user():
    print("Hello! Welcome to Python functions.")

# 2. Calling the function to execute it
greet_user()
```

### 📦 2. Function Arguments and Parameters
Functions can accept input variables to process dynamic data layouts:
* **Parameter:** The variable listed inside the function definition parentheses (the placeholder).
* **Argument:** The actual concrete value sent to the function when it is called.

```python
# 'name' is the input parameter (the placeholder variable)
def welcome_member(name):
    print(f"Welcome to the portal, {name}!")

# Passing unique string arguments into the function call
welcome_member("Alice")
welcome_member("Bob")
```

```python
def hi_function(name = "Joe"):
    print("Hi " + name)
    
hi_function()
```

### 📋 3. Inserting a Collection as an Argument
You can send any data type into a function as an argument, including mutable collection objects like **Lists**. The function will receive and loop through the collection structure seamlessly.

```python
# Defining a function that processes data sequences
def loop_food_basket(food_list):
    for item in food_list:
        print(f"Processing basket item: {item}")

# Creating a standard list collection
myfruits = ["fig", "apple", "banana"]

# Sending the entire list container as a single argument
loop_food_basket(myfruits)
```

```python
def fruits(name):
    for i in name:
        print(i)

fruit_names = ["Apple","Banana", "Mango"]
fruits(fruit_names)
```

### 📤 4. Return and the Expression
To let a function send a calculated data result back to the main script runner, use the **`return`** statement. 
* **Crucial Rule:** `return` exits the function immediately. Any code written beneath it inside the function block will be completely ignored!

```python
# Defining a function that calculates a math operation
def multiply_by_five(number):
    return number * 5

# Saving the returned output value into a variable
result = multiply_by_five(12)
print("The calculated return value is:", result)
```

```python
def degrees(x):
    return 15 * x

print(degrees(2))
print(degrees(3))
```

### 👥 6. Functions with Multiple Arguments

In Python, you can write reusable blocks of code called functions using the `def` keyword. Functions can accept multiple variables (known as parameters) separated by commas, allowing you to pass multiple pieces of information into them when they are called.


```python
# Define a function that accepts three distinct arguments
def calculate_trip_cost(fuel_price, distance, maintenance_fee):
    # Perform calculations using all incoming arguments
    fuel_cost = distance * fuel_price
    total_cost = fuel_cost + maintenance_fee
    return total_cost

# Call the function by passing three specific values in order
trip_total = calculate_trip_cost(1.5, 350, 45)
print(f"Total calculated trip cost: ${trip_total}")
```

```python
def new_names(name1, name2, name3):
    print("The first name is " + name1)
    print("The second name is " + name2)
    print("The last name is " + name3)

new_names(name1 = "Anna", name2 = "Jenny", name3 = "Tom")
```

### 👥 7. Arbitrary Arguments (`*args`)

When you want to define a function that can accept an unknown or flexible number of positional arguments, use the `*` symbol followed by a parameter name (conventionally named `args`). Python automatically packs all incoming stray arguments into a single, iterable **Tuple**.


```python
def fruits(*name):
    print("The fruit is " + name[1])
    print("The fruit is " + name[2])
    
fruits("Apple", "Banana", "Cherry")
```

```python
# Define a function that accepts any number of food ingredients
def make_sandwich(*args):
    print(f"Arguments packed into a tuple: {args}")
    print("Building sandwich with:")
    
    # Iterate through the tuple elements
    for ingredient in args:
        print(f" - {ingredient}")

# Call the function with 2 arguments
make_sandwich("Turkey", "Cheese")
print("-" * 30)

# Call the exact same function with 4 arguments
make_sandwich("Bacon", "Lettuce", "Tomato", "Mayo")
```

### 👥 8. Recursive Functions

A recursive function is a function that solves a problem by calling itself inside its own code definition. 

Every valid recursive function must have two critical parts:
1. **Base Case**: A specific condition that stops the function from calling itself. Without this, the function will loop infinitely and crash your program with a `RecursionError` (Stack Overflow).
2. **Recursive Case**: The part of the function where it calls itself, slightly changing the arguments each time to move closer toward the base case condition.


```python
def numbers(n):
    if(n>0):
        result = n + numbers(n-1)
        print(result)
    else:
        result = 0
    return result

print("\n \n recursion results")
numbers(6)
```

```python
# A simple recursive function to count down to zero
def countdown(n):
    # 1. Base Case: Stop when we reach 0
    if n <= 0:
        print("Blast off!")
        return
    
    # Action item
    print(n)
    
    # 2. Recursive Case: Call itself with n-1 (moving closer to 0)
    countdown(n - 1)

# Execution test
countdown(3)
```

### 👥 9. Introduction to Lambda (Anonymous) Functions

A **Lambda function** is a small, anonymous function that is defined without a name using the `lambda` keyword instead of the standard `def` keyword. 

📌 **Core Technical Characteristics:**
* **Anonymous Function**: They are nameless functions typically used for short, one-time operations.
* **Any Number of Arguments**: They can accept zero, one, or multiple arguments separated by commas.
* **Single Expression**: They can strictly only contain a single expression. They automatically return the result of that expression without using an explicit `return` statement.

📁 **Syntax Structure:**
`lambda arguments : expression`


```python
x = lambda i: i + 10
print(x(5))
```

```python
# 1. Basic single-argument lambda function (Squaring a number)
square = lambda x: x ** 2
print(f"Square of 5 using lambda: {square(5)}")

# 2. Multi-argument lambda function (Adding three numbers together)
add_three = lambda a, b, c: a + b + c
print(f"Sum of multiple arguments: {add_three(10, 20, 30)}")
```

```python
calculate = lambda x, y : x* y
print(calculate(5, 7))
```

```python
math_op = lambda x, y, z : x * y + z
print(math_op(2, 3, 4))
```

```python
# Lambda with user fucntion
def multiply(n):
    return lambda x: x * n

double = multiply(2)
triple = multiply(3)

print(double(10))
print(triple(10))
```
