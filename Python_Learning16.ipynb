# 📊 Introduction to Matplotlib

**Matplotlib** is a comprehensive, low-level data visualization library in Python used for creating static, animated, and interactive visualizations.

📌 **Core Capabilities of Matplotlib:**
* **Used to Build Plots**: It provides tools to generate publication-quality figures, charts, and diagrams from data arrays and dataframes.
* **Dimensional Versatility**: It natively supports the creation of standard 2D charts, complex 3D data projections, and dynamic multi-frame animations.

### 🔍 1. Importing Matplotlib and Checking Library Version

To use Matplotlib, you bring the core package into your active session using the `import matplotlib` command statement. You can inspect the version string asset of the library using the built-in `.__version__` attribute to ensure compatibility with your scripts.

### 📈 2. Introduction to Line Plots with Pyplot

Most of the utilities in Matplotlib are located under the `pyplot` submodule. It is standard convention to import it using the short alias `plt`. A line plot connects data sequence coordinates with straight line segments, making it ideal for tracking changes over continuous intervals.


```python
import matplotlib.pyplot as plt
import numpy as np

# Define coordinate arrays
xpoints = np.array([1, 2, 6, 8])
ypoints = np.array([3, 8, 1, 10])

# Draw a line plot connecting the coordinate points
plt.plot(xpoints, ypoints)

# Display the rendering output screen
plt.show()
```

### 🌊 3. Plotting Mathematical Functions (Sine Wave Line Plot)

You can plot continuous mathematical curves by using `np.arange(start, stop, step)` to generate a dense sequence of values, and passing that array into a mathematical function like `np.sin()`. The magic command `%matplotlib inline` ensures that the plots render directly below your code cell within the notebook interface.

```python
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline

# Generate values from 0 to 4*pi with a step sequence of 0.1
x = np.arange(0, 4 * np.pi, 0.1)
y = np.sin(x)

# Render the continuous curve line plot
plt.plot(x, y)
plt.show()
```

### 💾 4. Enhancing Graph Labels and Saving Plots Globally

You can make your plots much easier to read by using labels and titles. Use `plt.xlabel()`, `plt.ylabel()`, and `plt.title()` to add text annotations to your graph's axes and layout headers. You can also assign descriptions to your lines using `plt.legend()`.

To export your completed visualization as an image file on your system, use the `plt.savefig()` function.

📌 **Key Arguments Used in `plt.savefig()`:**
* **`dpi` (Dots Per Inch)**: Sets the image resolution. A higher value (like 350) means a sharper image.
* **`bbox_inches='tight'`**: Trims unnecessary whitespace from the edges so labels aren't accidentally cut off.

```python
import matplotlib.pyplot as plt
%matplotlib inline

# Coordinate arrays
x = [0, 2, 4, 6]
y = [1, 3, 4, 8]

# Create line plot
plt.plot(x, y)

# Add chart metadata labels
plt.xlabel('x values')
plt.ylabel('y values')
plt.title('plotted x and y values')
plt.legend(['Data 1'])

# Export plot to system storage as a PNG file
plt.savefig('plot1.png', dpi=350, bbox_inches='tight')

# Display plot rendering output
plt.show()
```

### 🔀 5. Plotting Multi-Line Trends using Object-Oriented Subplots

To plot multiple data lines onto a single chart layer, you can use Matplotlib's object-oriented approach by calling `plt.subplots()`. This returns a figure container (`fig`) and a specific axes layout layer (`ax`). Calling `.plot()` multiple times sequentially on the `ax` object overlays your data curves on top of each other automatically inside the same graph window.

```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

# Generate sequential array coordinates
x = np.arange(0, 4 * np.pi, 0.1)
y = np.sin(x)
z = np.cos(x)

# Initialize the object-oriented canvas grid layer
fig, ax = plt.subplots()

# Overlay independent data curves onto the same axes frame layer
ax.plot(x, y)
ax.plot(x, z)

# Render plot canvas
plt.show()
```

### 🏷️ 6. Annotating Subplots Using Object-Oriented Methods

When using the object-oriented subplot workflow (`fig, ax = plt.subplots()`), you apply graph metadata annotations using direct method calls on the axes variable instance (`ax`). 

📌 **Object-Oriented Labeling Methods:**
* **`ax.set_title()`**: Sets the chart's main header title.
* **`ax.legend()`**: Generates a color-coded key to identify multiple lines.
* **`ax.xaxis.set_label_text()`**: Assigns a label description below the horizontal X-axis.
* **`ax.yaxis.set_label_text()`**: Assigns a label description beside the vertical Y-axis.

```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

# Generate sequential array coordinates
x = np.arange(0, 4 * np.pi, 0.1)
y = np.sin(x)
z = np.cos(x)

# Initialize the object-oriented canvas grid layer
fig, ax = plt.subplots()

# Overlay independent data curves onto the same axes frame layer
ax.plot(x, y)
ax.plot(x, z)

# Apply metadata annotations using object-oriented axes methods
ax.set_title("Two trig functions")
ax.legend(['sin', 'cos'])
ax.xaxis.set_label_text('Angle')
ax.yaxis.set_label_text('Sine and Cosine')

# Render plot canvas
plt.show()
```

### 📊 7. Visualizing Data Distributions Using Histograms

A **Histogram** is used to visualize the frequency distribution of a continuous dataset. It breaks the data down into intervals called **bins** and displays the count of values falling into each bin as a vertical bar.

📌 **Key Functions Used:**
* `plt.style.use('fivethirtyeight')`: Applies a specific pre-defined visual style sheet layout to the chart canvas.
* `np.random.normal(mu, sigma, size)`: Generates a data array containing values drawn from a normal (Gaussian) distribution.
* `ax.hist(x, 20)`: Renders the histogram chart using 20 distinct data bins.
* `fig.tight_layout()`: Automatically adjusts subplot parameters to prevent text labels from overlapping.

```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
plt.style.use('fivethirtyeight')

mu = 80
sigma = 7
x = np.random.normal(mu, sigma, size=200)

fig, ax = plt.subplots()

ax.hist(x, 20)
ax.set_title("New Historgram")
ax.set_xlabel('bin range')
ax.set_ylabel('frequency')

fig.tight_layout()
plt.show()
```
