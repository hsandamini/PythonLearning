# 📂 Python File Handling: Opening and Accessing Files

Python provides built-in functions to safely interact with files stored on your operating system. The core function used to interact with any file structure is `open()`.

📁 **Syntax Structure:**
`open("filename", "mode")`

📌 **The Four Essential File Interaction Modes:**
1. **`r` (Read Mode)**: The default option. It opens a file for reading text. If the specified file does not exist, it raises a `FileNotFoundError`.
2. **`a` (Append Mode)**: Opens a file to add data to the end of its existing content. It automatically creates the file if it doesn't exist.
3. **`w` (Write Mode)**: Opens a file for writing. Note that it completely overwrites any data currently inside the file. It also creates a new file if it doesn't exist.
4. **`x` (Create Mode)**: Exclusively creates a brand new file. If a file with that name already exists anywhere in the directory path, the code crashes with a `FileExistsError`.


### 1. Read Files

```python
f  = open("demo.txt", "r")
print(f.read())
```

### 📖 2. Reading Specific Character Lengths

By passing an integer argument into the `.read(n)` method, you can specify exactly how many characters (or bytes) you want to extract from the file stream instead of loading the entire document into memory. Python starts reading from the current cursor position and counts whitespace and newline characters (`\n`).


```python
# Open the file in read mode matching your image structure
f = open("demo.txt", "r")

# Read and print only the first 7 characters
print(f.read(7))

# Always close the file cursor stream when finished
f.close()
```

### 📜 3. Reading Files Line by Line

The `.readline()` method reads a single line from the file stream at a time, stopping whenever it encounters a newline character (`\n`). Calling it multiple times sequentially moves the internal file pointer down to read the subsequent lines of the text document.


```python
# Open the target text document in read mode
f = open("demo.txt", "r")

# Read and output each line sequentially from top to bottom
print(f.readline())
print(f.readline())
print(f.readline())

# Close the active file object stream from memory
f.close()
```

### 🔄 4. Iterating Through a File Object with Loops

Instead of reading the entire file at once or manually typing multiple line statements, you can loop directly over the file handle using a standard `for` loop. This method processes the file lines line by line on-the-fly, making it highly efficient for reading extremely large text documents without exhausting system memory.


```python
# Open the specified file structure in standard read mode
f = open("demo.txt", "r")

# Loop over the file object to stream and print lines dynamically
for x in f:
    print(x)

# Terminate the file stream connection to clean up system state
f.close()
```

### ✍️ 5. Writing and Appending to Files

To modify a file or log new entries, you can use the write method along with an appropriate file mode parameter. 

📌 **Key Functional Differences:**
* **`"a"` (Append Mode)**: Adds new text content straight to the very end of the file. If the targeted file does not exist in the directory yet, Python automatically builds a new file for you without throwing an error.
* **`"w"` (Write Mode)**: Overwrites the existing file content entirely from scratch.


```python
# 1. Open a file in append mode ('a') to add data or create it if missing
f = open("main.txt", "a")
f.write("Another file is created and we are writing to it!")
f.close()

# 2. Open the newly modified file in read mode ('r') to verify the contents
f = open("main.txt", "r")
print(f.read())
f.close()
```

### 🆕 6. Creating and Initializing Files

You can use explicit modes to handle file creation behaviors depending on whether you want to prevent or allow overwriting existing assets.

📌 **Modes for Creating Files:**
* **`"x"` (Exclusive Creation)**: Creates a specified file. If a file with that name already exists anywhere in the directory path, Python crashes with a `FileExistsError`.
* **`"w"` (Write/Create)**: Creates a new empty file if it does not exist yet. If it does exist, it completely wipes out all old data to open a blank writing stream.


```python
# 1. Use 'x' mode to exclusively create a brand new file
try:
    f = open("file.txt", "x")
    f.close()
    print("Successfully created a new file with 'x' mode.")
except FileExistsError:
    print("File already exists!")

# 2. Use 'w' mode to create a new file or overwrite an existing one
f = open("new.txt", "w")
f.close()
print("File initialized or cleared using 'w' mode.")
```
